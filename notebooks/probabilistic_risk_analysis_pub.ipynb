# Probabilistic insurance risk

Reproduce the v1.1.0 publication results from the complete archive. See the repository README for download and installation instructions.

```python
from pathlib import Path
import subprocess
import sys
import pandas as pd
from IPython.display import Image, display

repo = Path.cwd()
if repo.name == 'notebooks':
    repo = repo.parent
archive = repo / 'results/campaign'
if not (archive / 'manifest.json').is_file():
    raise FileNotFoundError('Extract the v1.1.0 results archive into results/campaign first.')
```

```python
subprocess.run([sys.executable, str(repo / 'scripts/analysis/publication/reproduce.py'),
                '--archive', str(archive), '--out-dir', str(repo / 'results/publication')], check=True)
```

## Figures

The financial displays are generated from the archived season-level outcomes. Figure S1 is the published reference comparison; its event-level inputs are restricted.

```python
for figure in ['fig1_systemic_risk_overview_florida', 'fig_public_burden_scaling_linear', 'fig_combined_climate_policy_systemic_risk', 'fig_climate_buildingcode_sensitivity_public_burden', 'fig_loss_return_period', 'fig_climate_buildingcode_sensitivity', 'fig_combined_climate_policy_systemic_risk_ssp585', 'fig_residual_financing_severity_decomposition']:
    display(Image(filename=str(repo / 'results/publication/figures' / (figure + '.png'))))
```
